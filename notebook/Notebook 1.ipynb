# Exercício 02

```python
# Exercício 02
"""
Uma empresa de tecnologia está monitorando a quantidade de acessos diários ao seu site nos últimos 30 dias. 
Eles desejam realizar uma simulação para prever o comportamento do número de acessos nos próximos dias. 
Para isso, você deve:

1) Criar uma função chamada gerar_acessos que simula a quantidade de acessos diários ao site, 
retornando uma lista de números inteiros aleatórios entre 100 e 500, representando os acessos em 30 dias.
2) Com os dados gerados pela função, escreva uma nova função chamada calcular_media_acessos que 
calcule e retorne a média de acessos durante o período de 30 dias.

Suponha que o site da empresa terá uma promoção no próximo mês e eles esperam que os acessos aumentem em 20%. 
Escreva uma função chamada prever_acessos_futuros que receba a lista de acessos gerados e retorne uma
nova lista com os acessos ajustados, considerando o aumento de 20%.

"""


def gerar_acessos(ultimos_dias=30):
    
    acessos = []
    
    while True:
        if len(acessos) < ultimos_dias:
            acessos.append(random.randint(100, 500))
        else:
            break
    
    return acessos


def calcular_media_acessos(acessos):
    
    total_acessos = sum(acessos)
    quantidade_acessos = len(acessos)
    
    media = total_acessos/quantidade_acessos
    
    return media


def prever_acessos_futuros(acessos, promoção=0.2):
    
    acessos_futuros = []
    
    for acesso in acessos:
        acesso = acesso + promoção * acesso
        acessos_futuros.append(acesso)
    
    return acessos_futuros


def main():
    acessos = gerar_acessos()
    media_acessos = calcular_media_acessos(acessos)
    acessos_futuros = prever_acessos_futuros(acessos)
    print(f'Os acessos durante o mês foram de: {acessos}')
    print(
        f"A média de acessos diários no último mês foi de {media_acessos:.2f} acessos."
    )
    print(f"Previsão de acessos para o próximo mês: {acessos_futuros}")


if __name__ == "__main__":
    main()
```

```python
# Exercício 03
"""
Você foi encarregado de realizar uma simulação para analisar o comportamento de um dado de seis faces 
(numerado de 1 a 6). Para isso, siga os passos abaixo:

Crie uma função chamada simular_lancamento_dado que simule o lançamento de um dado 100 vezes.
A função deve retornar uma lista com os resultados dos lançamentos, ou seja, números inteiros entre 1 e 6.

Usando os dados gerados pela função, responda às perguntas abaixo:

a) Qual foi o número que mais apareceu nos lançamentos?
b) Qual foi o número que menos apareceu?
c) Qual é a frequência de cada número (de 1 a 6) nos 100 lançamentos?
d) Qual foi a porcentagem de vezes que o número 6 apareceu nos lançamentos?

Escreva uma função analisar_resultados_dado que receba a lista de resultados dos lançamentos e
e mostre as respostas de cada um dos itens acima.

# Exemplo de saída esperada:

Lançamentos: [3, 6, 2, 1, 5, 4, 6, ..., 2]
Número que mais apareceu: 6
Número que menos apareceu: 1
Frequência de cada número: {1: 12, 2: 15, 3: 20, 4: 18, 5: 17, 6: 18}
Porcentagem de 6: 18%

"""


def simular_lancamento_dado():
    
    lancamentos = []
    
    while True:
        if len(lancamentos) < 100:
            lancamentos.append(random.randint(1,6))
        else:
            break
    
    return lancamentos

lancamentos = simular_lancamento_dado()


def analisar_resultados_dado(lancamentos):
    
    print(f'Lançamentos: {lancamentos}')
    
    lancamento_1 = []
    lancamento_2 = []
    lancamento_3 = []
    lancamento_4 = []
    lancamento_5 = []
    lancamento_6 = []
    
    quant_1, quant_2, quant_3, quant_4, quant_5, quant_6 = [0, 0, 0, 0, 0, 0]
        
    for lancamento in lancamentos:
        if lancamento == 1:
            lancamento_1.append(lancamento)
            quant_1 = math.fsum(lancamento_1)
        elif lancamento == 2:
            lancamento_2.append(lancamento)
            quant_2 = math.fsum(lancamento_2)/2
        elif lancamento == 3:
            lancamento_3.append(lancamento)
            quant_3 = math.fsum(lancamento_3)/3
        elif lancamento == 4:
            lancamento_4.append(lancamento)
            quant_4 = math.fsum(lancamento_4)/4
        elif lancamento == 5:
            lancamento_5.append(lancamento)
            quant_5 = math.fsum(lancamento_5)/5
        elif lancamento == 6:
            lancamento_6.append(lancamento)
            quant_6 = math.fsum(lancamento_6)/6
    
    quantidades = [quant_1, quant_2, quant_3, quant_4, quant_5, quant_6]
    
    frequencia = {i + 1: quantidades[i] for i in range(6)}
    porcentagem_6 = quant_6/len(lancamentos) * 100
    
    print(f'Número que mais apareceu: {quantidades.index(max(quantidades)) + 1}')
    print(f'O número que menos apareceu: {quantidades.index(min(quantidades)) + 1}')
    print(f'Frequência de cada número: {frequencia}')
    print(f'Porcentagem de 6: {porcentagem_6:.0f}%')    

    
def main():
    lancamentos = simular_lancamento_dado()
    analisar_resultados_dado(lancamentos)


if __name__ == "__main__":
    main()
```

```python
# Exercício 04
"""
Uma fábrica de M&M's está avaliando a qualidade de seus sacos de M&M's com base na distribuição de cores. 
O objetivo é garantir que nenhum saco tenha uma diferença muito grande entre a cor que mais aparece e a cor 
que menos aparece. As cores são: vermelho, azul, amarelo, verde, laranja, e marrom.

As regras de qualidade estabelecem que:
    A quantidade da cor que mais aparece não pode exceder 50% da quantidade total de M&M's no saco.
    A quantidade da cor que menos aparece também não pode ser inferior a 10% do total de M&M's.

Para ajudar a fábrica a avaliar a qualidade dos sacos, você deve criar uma função chamada 
analisar_cores_mms que retorna a quantidade de sacos que atendem às regras de qualidade. 
A função deve receber uma lista de sacos de M&M's, onde cada saco é representado por uma 
lista com a quantidade de cor de cada M&M. 
Por exemplo, o saco [10, 20, 30, 40, 50, 60] contém 10 M&M's vermelhos, 20 azuis, 30 amarelos, 
40 verdes, 50 laranjas e 60 marrons.

"""


def analisar_cores_mms(sacos):
    
    sacos_validos = []
    
    for saco in sacos:
        total_mms = sum(saco)
        if max(saco) <= 0.5 * total_mms and min(saco) >= 0.1 * total_mms:
            sacos_validos.append(saco)
    
    return sacos_validos


def main():
    # Gerar sacos aleatórios de M&M's com semente constante (não modificar)"
    import random

    random.seed(42)
    sacos = [[random.randint(1, 100) for _ in range(6)] for _ in range(100)]

    # Avaliar sacos de M&M's
    resultado = analisar_cores_mms(sacos)

    # Imprimir resultado
    print(f"Sacos de M&M's que atendem às regras de qualidade: {resultado}")


if __name__ == "__main__":
    main()
```

```python
# Exercício 05
"""O Instituto Nacional de Meteorologia (INMET) está monitorando as chuvas em diversas regiões do Brasil e 
gerou um conjunto de dados fictícios com a quantidade de chuva acumulada (em milímetros) nos últimos 12 meses 
em 5 capitais brasileiras: São Paulo, Rio de Janeiro, Brasília, Salvador, e Porto Alegre.

Os dados são representados em uma lista de listas onde cada sublista contém o nome da cidade 
e uma lista com as chuvas acumuladas em milímetros para cada mês (de janeiro a dezembro):

"""

chuvas = [
    ["São Paulo", [150, 200, 180, 140, 160, 170, 210, 220, 200, 180, 190, 210]],
    ["Rio de Janeiro", [180, 220, 210, 190, 160, 150, 240, 250, 230, 210, 200, 190]],
    ["Brasília", [100, 120, 130, 110, 90, 80, 60, 50, 70, 110, 120, 140]],
    ["Salvador", [220, 240, 230, 210, 200, 220, 230, 240, 250, 260, 250, 240]],
    ["Porto Alegre", [180, 200, 190, 170, 160, 150, 180, 190, 210, 220, 230, 210]],
]

"""
Com base nesses dados, responda às seguintes perguntas:

    Qual cidade teve a maior quantidade total de chuva acumulada no ano?
    Qual cidade teve a menor quantidade total de chuva acumulada no ano?
    Qual foi o mês mais chuvoso em São Paulo e qual foi a quantidade de chuva acumulada nesse mês?
    Qual é a média mensal de chuva em Salvador ao longo do ano?
    Qual é a cidade com a maior variação de chuva entre o mês mais seco e o mês mais chuvoso? Qual é essa variação?


Crie funções para responder a cada uma das perguntas acima. 
Em seguida, adicione na função main a chamada para cada uma das funções e imprima os resultados.
"""

def maior_chuva_acumulada_total(chuvas):
    
    chuvas_acumuladas = []
    
    for chuva in chuvas:
        soma_chuvas = sum((chuva[1]))
        chuvas_acumuladas.append(soma_chuvas)
    
    maior_chuva = chuvas_acumuladas.index(max(chuvas_acumuladas))
    
    cidade_maior_chuva = chuvas[maior_chuva][0]
    
    return cidade_maior_chuva

def menor_chuva_acumulada_total(chuvas):
    
    chuvas_acumuladas = []
    
    for chuva in chuvas:
        soma_chuvas = sum(chuva[1])
        chuvas_acumuladas.append(soma_chuvas)
    
    menor_chuva = chuvas_acumuladas.index(min(chuvas_acumuladas))
    
    cidade_menor_chuva = chuvas[menor_chuva][0]
    
    return cidade_menor_chuva

def mes_mais_chuvoso(chuvas):
    
    cidade = input('Digite o nome da cidade com os acentos necessários e iniciais maiúsculas: ')
    
    if cidade == 'São Paulo':
        cidade = chuvas[0]
    elif cidade == 'Rio de Janeiro':
        cidade = chuvas[1]
    elif cidade == 'Brasília':
        cidade = chuvas[2]
    elif cidade == 'Salvador':
        cidade = chuvas[3]
    elif cidade == 'Porto Alegre':
        cidade = chuvas[4]
    
    meses = ['Janeiro', 'Fevereiro', 'Março', 'Abril', 'Maio', 'Junho', 'Julho', 'Agosto', 'Setembro', 'Outubro', 'Novembro', 'Dezembro']
    
    mes_maxima_chuva = cidade[1].index(max(cidade[1]))
    
    mes = meses[mes_maxima_chuva]
    
    chuva_acumulada = max(cidade[1])
    
    return {
        'cidade': cidade[0],
        'mês': mes,
        'chuva acumulada': chuva_acumulada
    }

def media_chuva(chuvas):
    
    cidade = input('Digite o nome da cidade com os acentos necessários e iniciais maiúsculas: ')
    
    if cidade == 'São Paulo':
        cidade = chuvas[0]
    elif cidade == 'Rio de Janeiro':
        cidade = chuvas[1]
    elif cidade == 'Brasília':
        cidade = chuvas[2]
    elif cidade == 'Salvador':
        cidade = chuvas[3]
    elif cidade == 'Porto Alegre':
        cidade = chuvas[4]
    
    media = sum(cidade[1])/len(cidade[1])
    
    return {'cidade': cidade[0],
            'media': media}

def variacao_chuva(chuvas):
    
    variacoes = []
    
    for chuva in chuvas:
        variacao = max(chuva[1]) - min(chuva[1])
        variacoes.append(variacao)
    
    maximo_varicao_index = variacoes.index(max(variacoes))
    
    cidade_variacao = chuvas[maximo_varicao_index][0]
    
    return cidade_variacao

def main():
    # Chame aqui as funções que respondem às perguntas
    resultado_1 = mes_mais_chuvoso(chuvas)
    resultado_2 = media_chuva(chuvas)
    print(f'Cidade com maior quantidade total de chuva: {maior_chuva_acumulada_total(chuvas)}')
    print(f'Cidade com menor quantidade total de chuva: {menor_chuva_acumulada_total(chuvas)}')
    print(f'O mês mais chuvoso em {resultado_1['cidade']} foi em {resultado_1['mês']} com acúmulo de {resultado_1['chuva acumulada']} milímetros de chuva')
    print(f'A média mensal de chuva em {resultado_2['cidade']} é de {resultado_2['media']}')
    print(f'A cidade com máxima variação de chuvas é {variacao_chuva(chuvas)}')


if __name__ == "__main__":
    main()
```

```python
# Exercício 07
"""
Você recebeu os dados do IPCA (Índice de Preços ao Consumidor Amplo) de diversas categorias de consumo 
em várias regiões brasileiras. Abaixo estão as variações percentuais no último mês para três grupos de consumo: 
Alimentação e bebidas, Habitação, e Vestuário, nas regiões Aracaju, Belo Horizonte, Belém, Brasília, Campo Grande
e Curitiba.

Os dados estão organizados da seguinte forma:
"""

cidades = ["Aracaju", "Belo Horizonte", "Belém", "Brasília", "Campo Grande", "Curitiba"]
alimentos_bebidas = [-1.04, 0.95, -0.33, 0.5, 0.49, 0.64]  # Aracaju, BH, Belém, Brasília, Campo Grande, Curitiba
habitacao = [1.82, 1.51, 1.44, 1.80, 1.78, 2.06]  # Aracaju, BH, Belém, Brasília, Campo Grande, Curitiba
vestuario = [0.95, 0.17, 0.21, 0.18, 0.13, -0.24]  # Aracaju, BH, Belém, Brasília, Campo Grande, Curitiba

""""
Com base nesses dados, responda às seguintes perguntas:

    Qual foi a maior variação no grupo "Alimentação e bebidas" entre todas as regiões?
    Qual foi a média de variação do grupo "Habitação" em todas as regiões?
    Quantas regiões apresentaram uma variação negativa no grupo "Vestuário"?
    Qual grupo apresentou a maior variação total somando todas as categorias?

Crie funções para responder a cada uma das perguntas acima.
"""

def maior_variacao(grupo_consumo):
    
    maior_variac = max(grupo_consumo)
    
    return maior_variac

def media_variacao(grupo_consumo):
    
    media = sum(grupo_consumo)/len(grupo_consumo)
    
    return media

def variacao_negativa(grupo_consumo):
    
    variacao_negativa = []
    
    for variacao in grupo_consumo:
        if variacao < 0:
            variacao_negativa.append(variacao)
    
    resultado = len(variacao_negativa)
    
    return resultado

def maior_variacao_total(grupo_1, grupo_2, grupo_3):
    
    grupos = [
        ['Alimentos e Bebidas', grupo_1],
        ['Habitação', grupo_2],
        ['Vestuário', grupo_3]
    ]
    
    somas = []
    
    soma_1 = sum(grupo_1)
    somas.append(soma_1)
    soma_2 = sum(grupo_2)
    somas.append(soma_2)
    soma_3 = sum(grupo_3)
    somas.append(soma_3)
    
    maximo_index = somas.index(max(somas))
    
    return grupos[maximo_index][0]


def main():
    # Chame aqui as funções que respondem às perguntas
    print(maior_variacao(alimentos_bebidas))
    print(media_variacao(habitacao))
    print(variacao_negativa(vestuario))
    print(maior_variacao_total(alimentos_bebidas, habitacao, vestuario))


if __name__ == "__main__":
    main()
```

```python
"""
Escreva uma função chamada meu_split que deve receber uma string e um delimitador.
A função deve dividir a string em partes com base no delimitador fornecido e retornar 
uma lista contendo as partes resultantes. A função deve replicar o comportamento básico
da função embutida split() do Python.
Requisitos
 • A função deve aceitar dois parâmetros:
    • texto (tipo str): A string que será dividida.
    • delimitador (tipo str): O caractere ou string que será usado como delimitador para dividir texto.
 • A função deve levar em consideração:
    • Delimitadores consecutivos (ou seja, se houver dois delimitadores seguidos, a função deve incluir na lista partes vazias).
    • Caso o delimitador não seja encontrado na string, a função deve retornar uma lista contendo a string original como seu único elemento.

Obs: Não utilize a função split do python nessa questão.
"""

def split(texto, delimitador):
    
    partes = []
    atual = ''
    i = 0
    
    if delimitador == '':
        raise ValueError('Erro: o delimitador não pode ser uma string vazia')
    
    while i < len(texto):
        
        if texto[i:i+len(delimitador)] == delimitador:
            partes.append(atual)
            atual = ''
            i += len(delimitador)
        else:
            atual += texto[i]
            i += 1
    
    partes.append(atual)
    
    return partes

def main():
    resultado = split("Olá, como você está?", " ")
    print(resultado)  # Saída esperada: ['Olá,', 'como', 'você', 'está?']

    resultado = split("Um|dois|três|quatro", "|")
    print(resultado)  # Saída esperada: ['Um', 'dois', 'três', 'quatro']

    resultado = split("Texto sem delimitador", "#")
    print(resultado)  # Saída esperada: ['Texto sem delimitador']


if __name__ == "__main__":
    main()
```
